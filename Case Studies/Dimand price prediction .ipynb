```python
import pandas as pd 
import numpy as np
import sklearn
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("/Users/pruth/Downloads/diamonds.csv")
```

```python
df
```

```python
x,xt,y,yt=train_test_split(df[['carat', 'depth', 'table', 'x', 'y',
       'z', 'cut', 'color', 'clarity']],df["price"],train_size=0.8,random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_t=pd.DataFrame(scaler.fit_transform(x[["carat","depth","table","x","y","z"]]),columns=scaler.get_feature_names_out())
xt_t=pd.DataFrame(scaler.fit_transform(xt[["carat","depth","table","x","y","z"]]),columns=scaler.get_feature_names_out())
```

```python
cat=x[["cut","color","clarity"]]
```

```python
from sklearn.neighbors import KNeighborsRegressor
GNB=KNeighborsRegressor()
GNB.fit(x_t,y)
predict=GNB.predict(xt_t)
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(yt,predict))
```
