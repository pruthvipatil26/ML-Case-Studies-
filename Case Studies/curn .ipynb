```python
df.TotalCharges=df.TotalCharges.apply(lambda x : np.nan if x ==' ' else x)
```

```python
df.TotalCharges=df.TotalCharges.astype(float)
```

```python
df.isna().sum()
```

```python
df.TotalCharges.fillna(value=df.TotalCharges.mean(),inplace=True)
```
