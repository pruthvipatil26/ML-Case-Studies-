```python
import pandas as pd
import numpy as np
import nltk
from sklearn.model_selection  import train_test_split
```

```python
df=pd.read_csv("./Classification/NLP/email_data.xls")
```

```python
df
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import re
from sklearn.feature_extraction.text import CountVectorizer
```

```python
def clean(text):
    text=re.sub("[^a-zA-Z]"," ",text)
    text=text.lower()
    text=text.split()
    text=[x for x in text if x not in stopwords.words("english")]
    return " ".join(text)
```

```python
from nltk.stem import WordNetLemmatizer
W=WordNetLemmatizer()
def lemma(text):
    text=text.split()
    text=[W.lemmatize(x) for x in text ]
    return " ".join(text)
```

```python
%%time
df["clean_text"]=df.text.apply(lambda x : clean(x))
```

```python
%%time
df["lemma_text"]=df.clean_text.apply(lambda x : lemma(x))
```

```python
df
```

```python
xt,xtt,yt,ytt=train_test_split(df.clean_text,df.label,test_size=0.2,random_state=1)
```

```python
cv=CountVectorizer()
xt_t=cv.fit_transform(xt)
xtt_t=cv.transform(xtt)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
LR=LogisticRegression()
LR.fit(xt_t,yt)
```

```python
predict=LR.predict(xtt_t)
```

```python
from sklearn.metrics import classification_report
print(classification_report(ytt,predict))
```

```python
KNN=KNeighborsClassifier()
KNN.fit(xt_t,yt)
predict=KNN.predict(xtt_t)
print(classification_report(ytt,predict))
```

```python
DTC=DecisionTreeClassifier()
DTC.fit(xt_t,yt)
predict=DTC.predict(xtt_t)
print(classification_report(ytt,predict))
```
