```python
def preprocessing(text):
    text=re.sub("[^a-zA-Z]", " ", text)
    print("text after removing the extra :  ",text)
    text=text.lower()
    print("after lower :  ",text)
    text=text.split()
    print("after split :  ",text)
    text=[x for x in text if x not in stopwords.words("english")]
    print("after stopwords :  ",text)
    textp=[p.stem(x) for x in text]
    print("after stemming :  ",textp)
    textl=[L.lemmatize(x) for x in text]
    return pd.DataFrame([textp].len())
```
