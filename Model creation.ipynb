```python
X=df[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]]
y=df["Species"]
```

```python
#### 6.Performe Train test Split
from sklearn.model_selection import train_test_split
X_train, X_test , y_train,y_test = train_test_split(X,y, test_size = 0.25, random_state=0)
```

```python
#Standardlizing the data
from  sklearn.preprocessing import StandardScaler
scaler = StandardScaler()  #Initializing the standard scaler object

X_train_transformes = scaler.fit_transform(X_train)  #LEARN PARAMETESR #APPLY TENASFORMED
X_train_transformes.shape


X_test_transformed = scaler.transform(X_test)
X_test_transformed.shape
```

## Model Building

```python
# using the logistic Regression
from sklearn.linear_model import LogisticRegression
LR_classifier=LogisticRegression()
LR_classifier.fit(X_train_transformes,y_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
KNN_classifier=KNeighborsClassifier()
KNN_classifier.fit(X_train_transformes,y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
DT_classifier=DecisionTreeClassifier()
DT_classifier.fit(X_train_transformes,y_train)
```

```python
# random forest classfier
from sklearn.ensemble import RandomForestClassifier
RFC_classifier=RandomForestClassifier()
RFC_classifier.fit(X_train_transformes,y_train)
```

```python
# using SVM(SVC)
from sklearn.svm import SVC
SV_classifier=SVC()
SV_classifier.fit(X_train_transformes,y_train)
```

## Prediction and Evaluation

```python
from sklearn import metrics
```
