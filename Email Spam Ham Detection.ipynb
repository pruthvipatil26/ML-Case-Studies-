## Data Cleaning And PreProcessing

```python
from sklearn.model_selection import train_test_split
from nltk.corpus import stopwords
stopwords=stopwords.words("english")  # English stopwords
from nltk.tokenize import word_tokenize
import re
from nltk.stem import WordNetLemmatizer
lemma=WordNetLemmatizer()
```

```python
def clean(text) :
    text=re.sub("[^a-zA-Z]"," ",text)

    text=text.lower()

    text=word_tokenize(text,preserve_line=True)

    text=[x for x in text if x not in stopwords]

    test=[lemma.lemmatize(x) for x in text]

    return " ".join(text)
```

```python
df["clean_text"] = df.text.apply(lambda x : clean(x ))
```

```python
df.head()
```

##  Data Split 

```python
x_train,x_test,y_train,y_test=train_test_split(df.clean_text,df.label_num ,test_size=0.2,random_state=1)
```

```python
x_test.shape,x_train.shape
```

## Data Transformation 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
```

```python
x_train_trans=tfidf.fit_transform(x_train)
x_test_trans=tfidf.transform(x_test)
```

## Model Creation

```python
from sklearn.linear_model import LogisticRegression
LR=LogisticRegression()
LR.fit(x_train_trans,y_train)
```

```python
from sklearn.metrics import classification_report
```

```python
predict=LR.predict(x_test_trans)
print(classification_report(y_test,predict))
```

## SVM

```python
from sklearn.svm import SVC
svc=SVC()
svc.fit(x_train_trans,y_train)
predict=svc.predict(x_test_trans)
print(classification_report(y_test,predict))
```

## KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(x_train_trans,y_train)
predict=knn.predict(x_test_trans)
print(classification_report(y_test,predict))
```

## Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
tree=KNeighborsClassifier()
tree.fit(x_train_trans,y_train)
predict=tree.predict(x_test_trans)
print(classification_report(y_test,predict))
```
