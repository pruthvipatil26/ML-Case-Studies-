```python
y=df["price"]
x=df[['carat', 'cut', 'color', 'clarity', 'depth', 'table', 'x', 'y',
       'z']]
```

```python
from sklearn.preprocessing import LabelEncoder

# create independent encoders
le_cut = LabelEncoder()
le_clarity = LabelEncoder()
le_color = LabelEncoder()

# fit on training
cut = pd.DataFrame(le_cut.fit_transform(x_train_categorical["cut"]), columns=["cut"])
clarity = pd.DataFrame(le_clarity.fit_transform(x_train_categorical["clarity"]), columns=["clarity"])
ccolor = pd.DataFrame(le_color.fit_transform(x_train_categorical["color"]), columns=["color"])
a = pd.concat([cut, clarity, ccolor], axis=1)
contraint = pd.concat([a, x_train_numeric_trans], axis=1)

# transform test
cut = pd.DataFrame(le_cut.transform(x_test_categorical["cut"]), columns=["cut"])
clarity = pd.DataFrame(le_clarity.transform(x_test_categorical["clarity"]), columns=["clarity"])
ccolor = pd.DataFrame(le_color.transform(x_test_categorical["color"]), columns=["color"])
b = pd.concat([cut, clarity, ccolor], axis=1)
contest = pd.concat([b, x_test_numeric_trans], axis=1)
```

# KNN

```python
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor()
knn.fit(x_train_trans,y_train)
predict=knn.predict(x_test_trans)
print("mean abslute error",metrics.mean_absolute_error(y_test,predict))
print("mean_squared_error",metrics.mean_squared_error(y_test,predict))
print("root_mean_squared_error",metrics.root_mean_squared_error(y_test,predict))
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor()
dt.fit(x_train_trans,y_train)
predict=dt.predict(x_test_trans)
print("mean abslute error",metrics.mean_absolute_error(y_test,predict))
print("mean_squared_error",metrics.mean_squared_error(y_test,predict))
print("root_mean_squared_error",metrics.root_mean_squared_error(y_test,predict))
```
