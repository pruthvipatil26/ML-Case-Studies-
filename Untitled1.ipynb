```python
import nltk
import pandas as pd
import numpy as np
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
v=TfidfVectorizer()
```

```python
df=pd.read_csv("../Supervised/Classification/NLP/email_data.xls")
```

```python
df.drop("Unnamed: 0",axis=1,inplace=True)
```

```python
df.text=df.text.apply(lambda x : x.replace("Subject:",""))
```

```python
df
```

```python
df.info()
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.text,df.label_num,train_size=0.8)
```

```python
x_train
```

```python
import nltk
from nltk.corpus import stopwords
```

```python
from nltk.stem import WordNetLemmatizer
lemm=WordNetLemmatizer()
import re
```

```python
def tranformation(x):
    x=re.sub("[^a-zA-Z]"," ",x)
    x=x.lower()
    x=x.split()
    x=[i for i in x  if i not in stopwords.words("english")]
    x=[lemm.lemmatize(i) for i in x]
    return " ".join(x)
```

```python
xt_t=x_train.apply(lambda a : tranformation(a) )
```

```python
xt_t
```

```python
%%time
xtt_t=x_test.apply(lambda a : tranformation(a) )
```

```python
xtt_t=tranformation(""" Subject: URGENT: Verify Your Bank Account Now
Body: Dear Customer, your account has been temporarily limited.
Verify your information immediately at [link] to avoid permanent suspension.
Failure to act will result in account closure. """)
```

```python
xtt_t
```

```python
xt_t=v.fit_transform(xt_t)
```

```python
xtt_t=v.transform([xtt_t])
```

```python
xtt_t.toarray()
```

```python
from sklearn.linear_model import LogisticRegression
l=LogisticRegression()
```

```python
l.fit(xt_t,y_train)
```

```python
predict=l.predict(xtt_t)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,predict))
```
