## Image Enhancements

```python
img=Image.open("./Classification/original_imgs/quantumania_2.png")
```

```python
plt.imshow(img)
```

```python
from PIL import ImageEnhance
```

```python
plt.figure(figsize=(12,12))
color_inhance=ImageEnhance.Color(img)
plt.subplot(2,2,1)
plt.imshow(img)

plt.subplot(2,2,2)
plt.imshow(color_inhance.enhance(0))

plt.subplot(2,2,3)
plt.imshow(color_inhance.enhance(2))

plt.subplot(2,2,4)
plt.imshow(color_inhance.enhance(5))
```

```python
plt.imshow(color_inhance.enhance(10))
```

```python
plt.figure(figsize=(12,12))
color_contrast=ImageEnhance.Contrast(img)
plt.subplot(2,2,1)
plt.imshow(img)

plt.subplot(2,2,2)
plt.imshow(color_contrast.enhance(0))

plt.subplot(2,2,3)
plt.imshow(color_contrast.enhance(2))

plt.subplot(2,2,4)
plt.imshow(color_contrast.enhance(5))
```

```python
plt.figure(figsize=(12,12))
color_contrast=ImageEnhance.Brightness(img)
plt.subplot(2,2,1)
plt.imshow(img)

plt.subplot(2,2,2)
plt.imshow(color_contrast.enhance(1))

plt.subplot(2,2,3)
plt.imshow(color_contrast.enhance(2))

plt.subplot(2,2,4)
plt.imshow(color_contrast.enhance(5))
```

```python
plt.figure(figsize=(12,12))
color_contrast=ImageEnhance.Sharpness(img)
plt.subplot(2,2,1)
plt.imshow(img)

plt.subplot(2,2,2)
plt.imshow(color_contrast.enhance(0))

plt.subplot(2,2,3)
plt.imshow(color_contrast.enhance(2))

plt.subplot(2,2,4)
plt.imshow(color_contrast.enhance(5))
```

## Filters

```python
from PIL import ImageFilter

plt.figure(figsize=(12,12))
plt.subplot(3,3,1)
plt.imshow(img)

plt.subplot(3,3,2)
plt.imshow(img.filter(ImageFilter.BLUR))

plt.subplot(3,3,3)
plt.imshow(img.filter(ImageFilter.CONTOUR))

plt.subplot(3,3,4)
plt.imshow(img.filter(ImageFilter.SMOOTH))

plt.subplot(3,3,5)
plt.imshow(img.filter(ImageFilter.SMOOTH_MORE))

plt.subplot(3,3,5)
plt.imshow(img.filter(ImageFilter.EDGE_ENHANCE))

plt.subplot(3,3,6)
plt.imshow(img.filter(ImageFilter.GaussianBlur))
```

## text font

```python
from PIL import Image,ImageDraw,ImageFont
img=Image.new(mode="RGB",size=(200,100),color=(1,1,1))
draw=ImageDraw.Draw(img)
text="Pruthviraj"
font=ImageFont.truetype("./Classification/Fonts/RubikDistressed-Regular.ttf",20)
font2=ImageFont.truetype("./Classification/Fonts/Pacifico-Regular.ttf",20)
font3=ImageFont.truetype("./Classification/Fonts/Inconsolata-Bold.ttf",20)
draw.text((50,10),text,font=font,fill="orchid")
draw.text((55,30),"Hanumant",font=font2,fill="royalblue")
draw.text((75,65),"Patil",font=font3,fill="lightsalmon")
plt.imshow(img)
```
