```python
# train test split
from sklearn.model_selection import train_test_split
xt,xtt,yt,ytt=train_test_split(x,y,train_size=0.80,random_state=0)
```

```python
from sklearn.linear_model import LogisticRegression
l=LogisticRegression()
l.fit(xt,yt)
predict=l.predict(xtt)
from sklearn.metrics import accuracy_score,classification_report
print(accuracy_score(ytt,predict))
print("---*"*20)
print(classification_report(ytt,predict))
```

```python
from sklearn.tree import DecisionTreeClassifier
tree=DecisionTreeClassifier()
tree.fit(xt,yt)
predict=tree.predict(xtt)
print(classification_report(ytt,predict))
```

```python
from sklearn.svm import SVC
svm=SVC()
svm.fit(xt,yt)
predict=svm.predict(xtt)
print(classification_report(ytt,predict))
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(xt,yt)
predict=knn.predict(xtt)
print(classification_report(ytt,predict))
```

```python
from sklearn.ensemble import BaggingClassifier
bc=BaggingClassifier()
bc.fit(xt,yt)
predict=bc.predict(xtt)
print(classification_report(ytt,predict))
```

```python
from sklearn.ensemble import AdaBoostClassifier
abc=AdaBoostClassifier()
abc.fit(xt,yt)
predict=abc.predict(xtt)
print(classification_report(ytt,predict))
```

```python
from sklearn.ensemble import ExtraTreesClassifier
etc=BaggingClassifier()
etc.fit(xt,yt)
predict=etc.predict(xtt)
print(classification_report(ytt,predict))
```
